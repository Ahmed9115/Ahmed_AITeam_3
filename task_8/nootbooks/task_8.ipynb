# third and forth part of the task : logistic regresion using pure python script !

```python
log = logistic_regression(10000 , 0.1)
```

```python
data = pd.read_csv('truth_table.csv')
y = data['Out']
y= np.reshape(y , (y.shape[0] ,1 ))
x = data.drop('Out' , axis= 1)
x_train , x_test , y_train, y_test = train_test_split(x , y ,test_size=0.2, random_state=42)
```

```python
log.fit(x_train,y_train)
log.predict(x_test)
log.validate(y_test)
```
