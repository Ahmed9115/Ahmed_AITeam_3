```python
import joblib
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import f1_score
```

```python
data = joblib.load("../processed_data.joblib")
```

```python
RBF_svm = SVC(kernel='rbf' , gamma='scale')
RBF_svm.fit(data['x_train'] , data['y_train'])
pred = RBF_svm.predict(data['x_test'])
print(f1_score(data['y_test'] , pred , average='macro'))
```
